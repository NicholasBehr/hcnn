```python
import jax
import jax.numpy as jnp


def h(x):
    return jnp.cos(x)


def h_sq(x):
    return jnp.conjugate(h(x)).T @ h(x)


def lin_prob(x):
    value = h_sq(x)
    jac = jax.jacobian(h_sq)(x)
    return jac.T, jac.T @ x - value


x = jnp.array([1.0, 0.0])
```

```python
value = h_sq(x)
jac = jax.jacobian(h_sq)(x)
jac.T
```

```python
a, b = lin_prob(x)
b
```

```python
from jax.numpy.linalg import solve as jsolve

jsolve(a, b)
```
