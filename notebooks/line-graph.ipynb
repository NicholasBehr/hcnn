```python
import networkx as nx

n = 10
node_graph = nx.Graph()
node_graph.add_nodes_from(range(0, n))

for x in range(0, n):
    for y in range(x, n):
        if x != y:
            node_graph.add_edge(x, y)
nx.draw_kamada_kawai(node_graph, with_labels=True)
# complete graph, #edges = n * (n-1) / 2 (wikipedia)
```

```python
adj = nx.adjacency_matrix(node_graph).toarray()
# print(adj)
print(adj.shape)
```

```python
nx.diameter(node_graph)
```

```python
edge_graph = nx.line_graph(node_graph)
nx.draw_kamada_kawai(edge_graph, with_labels=True)
```

```python
adj = nx.adjacency_matrix(edge_graph).toarray()
# print(adj)
print(adj.shape)  # n * (n-1) / 2
```

```python
nx.diameter(edge_graph)
```
