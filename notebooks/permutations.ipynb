```python
import itertools

vertices = ["A+", "A-", "B+", "B-", "C+", "C-", "D+", "D-"]
num = 0
for per in itertools.permutations(vertices):
    if per.index("A+") > per.index("A-"):
        continue
    if per.index("B+") > per.index("B-"):
        continue
    if per.index("C+") > per.index("C-"):
        continue
    if per.index("D+") > per.index("D-"):
        continue

    num += 1

    # print(per)

print(num)
```
