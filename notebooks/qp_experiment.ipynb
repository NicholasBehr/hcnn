```python
import jax.numpy as jnp
from jaxopt import BoxOSQP

q_mat = jnp.eye(2)
cost_vec = jnp.zeros((2,))
constr_mat = jnp.array([[1.0, 1.0]])
lower = jnp.array([1.0])
upper = jnp.array([jnp.inf])

qp = BoxOSQP()
sol = qp.run(params_obj=(q_mat, cost_vec), params_eq=constr_mat, params_ineq=(lower, upper)).params

print("primal", sol.primal)
print("dual eq", sol.dual_eq)
print("dual ineq", sol.dual_ineq)
```
